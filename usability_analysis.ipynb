## 6. Agreement Metrics

Calculate how well users and observers agree in their ratings.

```python
# Calculate agreement metrics
agreement = analyzer.calculate_agreement_metrics()
print("Agreement Metrics:")
print("=" * 50)
for metric, value in agreement.items():
    if 'Agreement' in metric or 'Correlation' in metric:
        print(f"{metric}: {value:.2f}%" if 'Agreement' in metric else f"{metric}: {value:.3f}")
    else:
        print(f"{metric}: {value:.3f}")

# Visualize agreement
agreement_df = agreement.to_frame(name='Value')
agreement_df
```

## 7. Detailed Question Analysis

Break down the responses question by question.

```python
# Create detailed comparison table
comparison = pd.DataFrame({
    'Question': analyzer.user_data['Question'].values,
    'User Score': analyzer.user_data['Score'].values,
    'Observer Score': analyzer.observer_data['Score'].values,
    'Difference': analyzer.user_data['Score'].values - analyzer.observer_data['Score'].values,
    'Abs Difference': np.abs(analyzer.user_data['Score'].values - analyzer.observer_data['Score'].values)
})

print("Detailed Question-by-Question Comparison:")
print(comparison)

# Highlight questions with largest discrepancies
print("\nQuestions with Largest Discrepancies:")
print(comparison.nlargest(3, 'Abs Difference')[['Question', 'User Score', 'Observer Score', 'Difference']])
```

## 8. Correlation Analysis

Visualize the relationship between user and observer scores.

```python
# Scatter plot with regression line
plt.figure(figsize=(10, 6))

user_scores = analyzer.user_data['Score'].values
observer_scores = analyzer.observer_data['Score'].values

# Scatter plot
plt.scatter(user_scores, observer_scores, s=100, alpha=0.6, edgecolors='black')

# Add regression line
z = np.polyfit(user_scores, observer_scores, 1)
p = np.poly1d(z)
plt.plot(user_scores, p(user_scores), "r--", linewidth=2, label=f'Trend line (y={z[0]:.2f}x+{z[1]:.2f})')

# Add diagonal line (perfect agreement)
plt.plot([0, 10], [0, 10], 'k-', linewidth=1, alpha=0.3, label='Perfect Agreement')

# Add labels for each point
for i, q in enumerate(analyzer.user_data['Question'].values):
    plt.annotate(f'Q{q}', (user_scores[i], observer_scores[i]), 
                xytext=(5, 5), textcoords='offset points', fontsize=8)

plt.xlabel('User Score', fontsize=12)
plt.ylabel('Observer Score', fontsize=12)
plt.title('User vs Observer Score Correlation', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Calculate and display correlation
correlation = np.corrcoef(user_scores, observer_scores)[0, 1]
print(f"\nCorrelation coefficient: {correlation:.3f}")
```
